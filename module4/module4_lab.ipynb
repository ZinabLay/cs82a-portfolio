```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("sales_clean.csv")

print("Dataset shape:", df.shape)
display(df.head())
print(df.dtypes)
```

```python
print("PRICE")
print("Mean:", df["price"].mean())
print("Median:", df["price"].median())
print("Standard deviation:", df["price"].std())

print("\nQUANTITY")
print("Mean:", df["qty"].mean())
print("Median:", df["qty"].median())
print("Standard deviation:", df["qty"].std())
```

### Which Summary Would I Report?

I would use the median price of $37.53 because it gives a better idea of what a typical item costs. The mean is higher because a few items have much higher prices. I think the median is more useful for describing the prices in this dataset.

```python
df["price"].plot(kind="hist", bins=20)

plt.title("Distribution of Product Prices")
plt.xlabel("Price")
plt.ylabel("Frequency")
plt.show()
```

### Histogram Shape

The histogram is right-skewed. Most products have lower prices, while only a few products have much higher prices.

```python
df["revenue"] = df["price"] * df["qty"]

display(df.head())
```

```python
correlation = df.corr(numeric_only=True)

display(correlation)
```

```python
df.plot.scatter(x="price", y="revenue")

plt.title("Price vs Revenue")
plt.xlabel("Price")
plt.ylabel("Revenue")
plt.show()
```

```python
df.plot.scatter(x="qty", y="revenue")

plt.title("Quantity vs Revenue")
plt.xlabel("Quantity")
plt.ylabel("Revenue")
plt.show()
```

### What I Would Tell a Manager

Price has a strong positive relationship with revenue (r = 0.82), while quantity has a weaker positive relationship (r = 0.33). This suggests that higher-priced items are more closely related to higher revenue in this dataset. However, a few unusually high values may affect the results, and correlation does not always mean causation.
