```python
import pandas as pd

df = pd.read_csv("allbacks.csv")

print("Dataset shape:", df.shape)
display(df.head())
print(df.dtypes)
```

```python
cover_sorted = df["cover"].sort_values().reset_index(drop=True)

print(cover_sorted)
print("Median cover type:", cover_sorted.iloc[len(cover_sorted) // 2])
```

```python
print("Mean weight of all books:", df["weight"].mean())
```

```python
print("Mean weight by cover type:")
print(df.groupby("cover")["weight"].mean())
```

### What the Results Mean

The median cover type is paperback (pb). This means that paperback is the middle cover type when the books are sorted by cover type.

The average weight of all books is about 706.67 grams.

Hardback books have an average weight of about 796.43 grams, while paperback books average about 628.13 grams. This suggests that hardback books are heavier than paperback books in this dataset.
