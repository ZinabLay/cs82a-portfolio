```python
import pandas as pd

df = pd.read_csv("messy_sales.csv")

print(df.shape)
display(df.head(10))
print(df.dtypes)
```

## Problems I Found

1. I noticed that some prices are missing in the dataset.
2. Some dates are written in different formats, so they need to be changed to one format.
3. Some ZIP codes are missing the zero at the beginning. For example, 2134 should be 02134.

```python
df.isna().sum()
```

```python
median_price = df["price"].median()

print("Median price:", median_price)

df["price"] = df["price"].fillna(median_price)

print("Missing prices after cleaning:")
print(df["price"].isna().sum())
```

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
df = df.drop_duplicates()

print("Shape after removing duplicates:", df.shape)
```

```python
df["zip"] = df["zip"].astype(str).str.zfill(5)

print(df["zip"].head(10))
print(df["zip"].dtype)
```

```python
df["date"] = pd.to_datetime(
    df["date"], format="mixed"
).astype("datetime64[ns]")
```

```python
print(df["date"].dtype)
```

```python
print("Negative quantities:")
display(df[df["qty"] < 0])
```

## Negative Quantity Decision

I found 3 rows with negative quantities. I decided to keep them because they could represent returned products. Removing them might make the sales data less accurate. However, I cannot be sure they are returns because the dataset does not explain why the quantities are negative.

```python
df.to_csv("sales_clean.csv", index=False)

print("Cleaned CSV saved successfully!")
print("Final shape:", df.shape)
print("Missing values:")
print(df.isna().sum())
```

```python
print("Final shape:", df.shape)
print("Date type:", df["date"].dtype)
print("ZIP type:", df["zip"].dtype)
print("Missing values:")
print(df.isna().sum())
```

## Cleaning Log

I started with 300 rows and 6 columns. First, I checked the dataset and found 12 missing prices. I filled them using the median price of 37.53.

Next, I found 8 duplicate rows and removed them. After that, the dataset had 292 rows.

I changed the ZIP codes to text and added zeros at the beginning when needed. I also changed all dates to the same datetime format.

I found 3 negative quantities. I kept them because they might be returned products, and I did not want to remove data that could be important.

The final cleaned dataset has 292 rows and 6 columns with no missing values.
