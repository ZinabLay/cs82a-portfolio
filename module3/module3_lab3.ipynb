```python
import pandas as pd

df = pd.read_csv("bridges.csv")

print("Dataset shape:", df.shape)
display(df.head())
print(df.dtypes)
```

## My Assumption

For this lab, I assume that values outside the 1.5 × IQR range may be unusual and should be flagged for review. I will investigate flagged values before deciding whether they are errors. I will not automatically remove unusual values.

```python
q1 = df["length"].quantile(0.25)
q3 = df["length"].quantile(0.75)

iqr = q3 - q1

low = q1 - 1.5 * iqr
high = q3 + 1.5 * iqr

print("Q1:", q1)
print("Q3:", q3)
print("IQR:", iqr)
print("Lower limit:", low)
print("Upper limit:", high)

print("Missing length values:")
print(df["length"].isna().sum())

print("Length summary:")
print(df["length"].describe())
```

```python
print("Flagged rows:")
display(df[(df["length"] < low) | (df["length"] > high)])
```

### Handling Missing Values

There are 27 missing values in the length column. I decided to leave them as missing because I do not know the actual bridge lengths. I do not want to guess numbers that may be incorrect.

### Outlier Screening Results

I used the 1.5 × IQR rule to screen the length column. The upper limit was 3,500, and three bridges were flagged: E34 (4,558), E46 (4,000), and E91 (3,756). These values are unusually high compared with the other recorded lengths, but I cannot conclude that they are incorrect based on length alone.

## Conclusion

I used the 1.5 × IQR rule to identify possible outliers in the bridge length column. Three bridges were flagged for review: E34, E46, and E91. I kept these values because being unusually high does not automatically mean they are incorrect. I also left the 27 missing length values unchanged because their actual lengths are unknown.
