### Temperature Cutoffs

- Cold: Below 15°C
- Warm: From 15°C to 25°C, including both boundaries.
- Hot: Above 25°C.

```python
Temperatures = [10,16,23,28,14,22,30]

for temp in Temperatures:
    if temp < 15:
        category = "Cold"
    elif temp <= 25:
        category = "Warm"
    else:
        category = "Hot"

    print(temp, "°C:", category)
```

```python
test_temps = [14,15,25,26]

for temp in test_temps:
    if temp < 15:
        category = "Cold"
    elif temp <= 25:
        category = "Warm"
    else:
        category = "Hot"

    print (temp,"°C:", category )
```
