## Multiplication the Long Way

Course: CS 82A - Introduction to Data Science
Assignment: Module 2 Lab 2

```python
def multiply_for(a, b):
    result = 0

    for i in range(b):
        result = result + a

    return result
```

```python
print("7 x 6 =", multiply_for(7, 6))
print("5 x 4 =", multiply_for(5, 4))
print("8 x 3 =", multiply_for(8, 3))
```

```python
def multiply_while(a, b):
    result = 0
    count = 0

    while count < b:
        result = result + a
        count = count + 1

    return result
```

```python
print("7 x 6 =", multiply_while(7, 6))
print("5 x 4 =", multiply_while(5, 4))
print("8 x 3 =", multiply_while(8, 3))
```

### How the While Loop Stops

The while loop stops when the count is no longer less than b. Each time the loop runs, count increases by 1. When count reaches b, the condition becomes False, and the loop ends.

```python
for a, b in [(7, 6), (5, 4), (8, 3)]:
    print(
        "For:", multiply_for(a, b),
        "| While:", multiply_while(a, b),
        "| Match:", multiply_for(a, b) == multiply_while(a, b)
    )
```
